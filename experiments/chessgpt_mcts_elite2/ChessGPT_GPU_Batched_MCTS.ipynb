## Interpreting the results

`GPU_TIMED` reports the **actual completed MCTS backup count** for a 2.5- or 5-second UCI movetime search. `GPU_GO_NODES_200` runs `go nodes 200` without a time cap and verifies exactly 200 backups. The latter proves the node command works but **does not** demonstrate 200 nodes within a blitz move.

If the timed count is below 200, further implementation tuning or hardware acceleration is still required. Search only uses ChessGPT probabilities and Stockfish classical evaluation, not an externally coded move-ordering heuristic.
